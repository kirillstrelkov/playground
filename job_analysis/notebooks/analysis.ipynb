# Job Requirements Analysis
Load CSV from CLI output and compute frequency table.


```python
from pathlib import Path

import pandas as pd

csv_path = Path("../data/out/skills.csv")  # adjust if needed
df = pd.read_csv(csv_path)
df.head()
```

# Group by skill

```python
# Group by Skill and count
df["skill"] = df["skill"].str.lower()
df.groupby(["skill"]).size().reset_index(name="count").sort_values(
    by="count", ascending=False
).head(30)
```
